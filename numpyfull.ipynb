## 01. Introduction, Installation & Setup

**What is NumPy?**

**NumPy (Numerical Python)** is an open-source Python library designed for efficient numerical computing. It provides a powerful multidimensional array object called ndarray and functions for mathematical, statistical, and linear algebra operations.**

**Why use NumPy?**
- Faster numerical operations than typical Python loops over lists.
- Efficient memory representation for homogeneous numerical data.
- Support for multidimensional arrays.
- Vectorized mathematical and statistical calculations.
- Foundation for scientific computing, machine learning, and data analysis.

## 28. Statistical Testing and Model Evaluation

**Definition:** Statistical Testing and Model Evaluation covers the NumPy concepts and APIs needed to work with numerical arrays reliably. Learn the shape, dtype, and axis behavior of every operation, then verify its output with small examples.

**Key syntax:** `confusion counts from masks`

**QA application:** Calculate metrics independently from a model implementation.

**Practice:** Change inputs, inspect `shape` and `dtype`, and add a meaningful assertion.


```python
import numpy as np
y=np.array([1,0,1,0,1,0]);
p=np.array([1,0,0,0,1,1]); 
tp=np.sum((y==1)&(p==1)); 
fp=np.sum((y==0)&(p==1)); 
fn=np.sum((y==1)&(p==0)); 
precision=tp/(tp+fp); 
recall=tp/(tp+fn); 
print(tp,fp,fn,precision,recall)
```
